# LayerNorm Implementation

### layernorm公式

$$
\text{LayerNorm}(x) = \gamma \frac{x - \mathbb{E}[x]}{\sqrt{\text{Var}(x) + \epsilon}} + \beta
$$
where $\gamma$ and $\beta$ are learnable parameters, $\mathbb{E}[x]$ is the mean of $x$, $\text{Var}(x)$ is the variance of $x$, and $\epsilon$ is a small constant for numerical stability.

```python
import torch
import torch.nn as nn
# layernorm from scratch
class LayerNorm(nn.Module):
    def __init__(self, features, eps=1e-6):
        super().__init__()
        self.gamma = nn.Parameter(torch.ones(features))
        self.beta = nn.Parameter(torch.zeros(features))
        self.eps = eps

    def forward(self, x):
        mean = x.mean(-1, keepdim=True)
        std = x.std(-1, keepdim=True)
        return self.gamma * (x - mean) / (std + self.eps) + self.beta
```

```python
# test
x = torch.randn(2, 3)
layer_norm = LayerNorm(3)
output = layer_norm(x)
print(output.size())
```
